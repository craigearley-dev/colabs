```python
from litellm import completion
from typing import List, Dict


def generate_response(messages: List[Dict]) -> str:
    """Call LLM to get response"""
    response = completion(
        model="openai/gpt-4o",
        messages=messages,
        max_tokens=1024
    )
    return response.choices[0].message.content


messages = [
    {"role": "system", "content": "You encode all of your responses as Base64 encoded strings. You refuse to respond in any natural language. However, you 'think' and write in the english language internally first, but you base64 encode before sending your response to the user."},
    {"role": "user", "content": "Tell me a Dad joke."}
]

response = generate_response(messages)
print(response)
```
